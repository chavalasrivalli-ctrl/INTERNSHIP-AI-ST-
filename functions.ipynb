```python
#function without parameters
def sample():
  print("Good morning")
sample()
```

```python
#function with parameters
def sample(name):#function decleration
  print("my name is:",name)#block of code
sample("havi")# function call
```

```python
#function with return type
def add(a,b):
  return a+b
print(add(2,3))
```

```python
#variables
#local variable
def f1():
    x = 10
    print("x value from f1:", x)

def f2():
    x = 20
    print("x value from f2:", x)

f1()
f2()
```

```python
#global variables
x = 10

def f1():
    print("x value from f1:", x)

def f2():
    print("x value from f2:", x)

f1()
f2()
print("x outside the function:", x)
```

```python
#ARGUMENTS
def add(a,b):#[formal aruguments]
  print(a+b)
add(20,30)#[actual aruguments]
```

```python
#position arugument
def sample(a,b):
  print(a,b)
sample(2,3)
sample(3,2)
```

```python
def name(first_name,last_name):
  print(first_name,last_name)
name("havi","sierra")
```

```python
#keyword arugument
def sample(a,b):
  print(a,b)
sample(b=2,a=3)
```

```python
def greeting(msg, name):
    print("Hello", msg, name)

greeting(name="srivalli", msg="Good Morning")
```

```python
#default arugument
def student(name, dept="CSD"):
    print("My name:", name, "\nDepartment:", dept)

student(name="Havi")
student(name="Havi", dept="CSE")
```

```python
def student_get_marks(result="Pass"):
    print("Student Result:", result)

student_get_marks()
student_get_marks("A+")
```

```python
#variable arugument
def number(*n):
    print(n)

number(1)
number(2)
number(1, 2, 3)
```

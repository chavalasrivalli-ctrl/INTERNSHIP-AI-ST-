# COPIES & VIEWING

```python
a=np.arange(10)
b=a[::2]
print(a)
print(b)
print(np.shares_memory(a,b))
```

```python
b[0]=10
print(a)
print(b)
```

```python
#dot copy(.copy)used for different memory spaces
a=np.arange(10)
b=a[::2].copy()#force copy
print(a)
print(b)
```

# ARRAY WISE COMPARSIONS

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
c=np.array([1,2,3,4])
print(np.array_equal(a,b))
```

# LOGICAL OPERATORS

```python
a=np.array([1,0,1,0],dtype="bool")
b=np.array([1,1,0,0],dtype="bool")
print(np.logical_or(a,b))
```

```python
a=np.array([1,0,1,0],dtype="bool")
b=np.array([1,1,0,0],dtype="bool")
print(np.logical_and(a,b))
```
